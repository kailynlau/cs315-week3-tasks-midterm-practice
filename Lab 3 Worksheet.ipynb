```python
from bs4 import BeautifulSoup
import re

with open('bongsoon.html', 'r', encoding='utf-8') as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, 'html.parser')
```

Check that the parsing was successful:

```python
# print(soup)
soup.title
```

```python
desc = soup.find("meta")["content"]
print(desc)
```

<h2 id=sec1>Regex Extraction</h2>

```python
# Task 1A- extract the 4-digit release year
pattern_1A = re.compile(r'\d{4}')

for p in pattern_1A.finditer(desc):
    print(p)

pattern_1B = re.compile(r'\b\w{6}\b')
for p in pattern_1B.finditer(desc):
    print(p)
```

<h2 id=sec2>Exercise 2:</h2>

```python
synopsis = soup.find("div", class_="show-synopsis")
synopsis = synopsis.find("p").text
print(synopsis)
```

```python
pattern_2A = re.compile(r'\b[aeiou]\w*\b')
for p in pattern_2A.finditer(synopsis):
    print(p)
```

```python
pattern_2B = re.compile(r'\b[^\Wae]+\b', re.IGNORECASE)
for p in pattern_2B.finditer(synopsis):
    print(p)
```

<h2 id=sec3>Exercise 3</h2>

```python
url = soup.find(string="Viki").find_parent('a')["href"]
print(url)
```

```python
pattern_3a = re.compile(r'www\.(.+?\.com)')

for p in pattern_3a.finditer(url):
    print(p)
```

```python
# pattern_3b = re.compile(r'?q=(.*)%www\.(.+?\.com).*%3F(\d*)')

pattern_3b = re.compile(r'(https?).*www\.(.+?\.com).*tv%2F(\d*)')

for p in pattern_3b.finditer(url):
    print(p.group(1))
    print(p.group(2))
    print(p.group(3))
```

<h2 id=sec4>Exercise 5</h2>

```python
results = soup.find("div", class_="col-xs-8 col-sm-9 col-lg-9")

ratings, watchers, review = results.find_all("div", class_="hfs")

text = f"Your Rating: 0/10\n{ratings.text}\n{watchers.text}\n{review.text}"
print(text)
```

```python
pattern_5 = (
    r"Ratings:\s*(?P<rating>[\d.]+)/10\s*from\s*(?P<rating_users>[\d,]+)\s*users.*?"
    r"# of Watchers:\s*(?P<watchers>[\d,]+).*?"
    r"Reviews:\s*(?P<review_users>[\d,]+)\s*users"
)

match = re.search(pattern_5, text, re.S) # used AI to ask how to get lazy quantifiers to match across line breaks

if match:
    print(match.groupdict())
```
